```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['axes.titlesize'] = 14
pd.set_option('display.max_columns', None)

df = pd.read_csv('../data/raw/Loan Approval Prediction Dataset.csv')
print("Shape:", df.shape)
print("\nFirst 5 rows:")
df.head()
```

```python
print("=" * 60)
print("DATA TYPES")
print("=" * 60)
print(df.dtypes)

print("\n" + "=" * 60)
print("NUMERIC SUMMARY")
print("=" * 60)
df.describe().T
```

```python
print("Loan_Status distribution:")
print(df['Loan_Status'].value_counts())
print("\nPercentages:")
print(df['Loan_Status'].value_counts(normalize=True).round(4) * 100)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Count plot
df['Loan_Status'].value_counts().plot(kind='bar', ax=axes[0], color=['#2ecc71', '#e74c3c'])
axes[0].set_title('Loan Status - Counts')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=0)

# Pie chart
df['Loan_Status'].value_counts().plot(kind='pie', ax=axes[1], autopct='%1.1f%%',
                                       colors=['#2ecc71', '#e74c3c'], startangle=90)
axes[1].set_title('Loan Status - Proportions')
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()

# Interpretation
approved_pct = (df['Loan_Status'] == 'Approved').mean() * 100
print(f"\n>>> Approved: {approved_pct:.2f}%, Rejected: {100 - approved_pct:.2f}%")
print(">>> Class imbalance is moderate (~64/36). SMOTE will help.")
```

```python
cat_cols = df.select_dtypes(include='object').columns.tolist()
cat_cols.remove('Loan_Status')  # target, not a feature

print(f"Categorical columns ({len(cat_cols)}): {cat_cols}\n")

for col in cat_cols:
    print(f"--- {col} ---")
    print(df[col].value_counts())
    print()
```

```python
cat_cols = [c for c in df.columns if df[c].dtype == 'object' or str(df[c].dtype) == 'str']
cat_cols.remove('Loan_Status')
```

```python
fig, axes = plt.subplots(5, 2, figsize=(14, 20))
axes = axes.flatten()

plot_cols = [c for c in cat_cols if c != 'Application_ID']  # skip ID

for i, col in enumerate(plot_cols):
    ct = pd.crosstab(df[col], df['Loan_Status'], normalize='index') * 100
    ct.plot(kind='bar', stacked=True, ax=axes[i], color=['#e74c3c', '#2ecc71'])
    axes[i].set_title(f'{col} vs Loan_Status')
    axes[i].set_ylabel('Percentage')
    axes[i].legend(title='Status', loc='upper right', fontsize=8)
    axes[i].tick_params(axis='x', rotation=45)

# hide unused
for j in range(len(plot_cols), len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print(f"Numeric columns ({len(num_cols)}): {num_cols}\n")

fig, axes = plt.subplots(6, 4, figsize=(20, 22))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    axes[i].hist(df[col].dropna(), bins=40, color='steelblue', edgecolor='black', alpha=0.7)
    axes[i].set_title(col, fontsize=10)
    axes[i].set_ylabel('Frequency')

for j in range(len(num_cols), len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(6, 4, figsize=(20, 22))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    axes[i].boxplot(df[col].dropna(), vert=True)
    axes[i].set_title(col, fontsize=10)

for j in range(len(num_cols), len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()

print("\nOutlier counts (IQR method):")
for col in num_cols:
    q1, q3 = df[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    if outliers > 0:
        print(f"  {col}: {outliers} outliers ({outliers/len(df)*100:.2f}%)")
```

```python
import numpy as np
n_cols = len(num_cols)
n_rows = (n_cols + 3) // 4  # ceil division for 4 columns per row

fig, axes = plt.subplots(n_rows, 4, figsize=(20, n_rows * 4))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    axes[i].boxplot(df[col].dropna(), orientation='vertical')
    axes[i].set_title(col, fontsize=10)

for j in range(n_cols, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()

print("\nOutlier counts (IQR method):")
for col in num_cols:
    q1, q3 = df[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    if outliers > 0:
        print(f"  {col}: {outliers} outliers ({outliers/len(df)*100:.2f}%)")
```

```python
corr = df[num_cols].corr()

plt.figure(figsize=(18, 14))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=False, cmap='coolwarm', center=0,
            square=True, linewidths=0.5, cbar_kws={"shrink": 0.8})
plt.title('Correlation Matrix (Numeric Features)')
plt.tight_layout()
plt.show()

df_temp = df.copy()
df_temp['Loan_Status_Bin'] = (df_temp['Loan_Status'] == 'Approved').astype(int)
target_corr = df_temp[num_cols + ['Loan_Status_Bin']].corr()['Loan_Status_Bin'].drop('Loan_Status_Bin')
target_corr_abs = target_corr.abs().sort_values(ascending=False)

print("\nTop 15 features by |correlation| with target:")
print(target_corr_abs.head(15))

print("\nFull correlation with target (signed):")
print(target_corr.sort_values(ascending=False).round(4))
```

```python
corr_matrix = df[num_cols].corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

high_corr_pairs = [(col, row, upper.loc[row, col]) 
                   for col in upper.columns 
                   for row in upper.index 
                   if upper.loc[row, col] > 0.85]

print("Highly correlated feature pairs (>0.85):")
if high_corr_pairs:
    for c1, c2, val in high_corr_pairs:
        print(f"  {c1} ↔ {c2}: {val:.3f}")
    print("\n>>> Consider dropping one from each pair.")
else:
    print("  None found.")
```

```python
print("Collateral value counts:")
print(df['Collateral'].value_counts())
print()

for col in ['Property_Value', 'Collateral_Value', 'Loan_to_Value']:
    print(f"Missing {col} by Collateral status:")
    print(df.groupby('Collateral')[col].apply(lambda x: x.isnull().sum()))
    print()

print("=" * 60)
print("Other missing columns vs target:")
for col in ['Other_Income', 'Investments', 'Bank_Balance', 'Credit_Score',
            'Interest_Rate', 'Tax_Return_Filed']:
    print(f"\n--- {col} ---")
    print(f"Missing count: {df[col].isnull().sum()}")
    print(df.groupby(df[col].isnull())['Loan_Status'].value_counts())
```
